```python
import pandas as pd

# 1. Load the raw sales file
df = pd.read_excel("/content/final_electronics_store_sales_data.xlsx")
```

```python
df.describe()
```

```python
df.info()
```

```python
# 2. Convert Date_Time column to datetime format
df["Date_Time"] = pd.to_datetime(df["Date_Time"])
```

```python
df.info()
```

```python
# 3. Extract the year-month format (e.g., '2026-07')
df["month"] = df["Date_Time"].dt.to_period("M").astype(str)
```

```python
df.info()
```

```python
df.head()
```

```python
# 4. Group by month, Product_ID, and Product_Name, then calculate aggregations
aggregated_df = (
    df.groupby(["month", "Product_ID", "Product_Name"])
    .agg(
        sales_units=("Quantity", "sum"),  # Sum of quantity
        price=("Unit_Price", "mean"),  # Average unit price
    )

)
```

```python
# 5. Round the average price to 2 decimal places for clean formatting
# aggregated_df["price"] = aggregated_df["price"].round(2)

# 6. Save to the requested output CSV file name
aggregated_df.to_csv("aggregated_sales.csv", index=False)
```
